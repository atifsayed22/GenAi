## Task 5 : Using filter () : Filter prices 

```python
prices = [ 100 , 250 , 400 , 1200 , 50 , 2000 , 850]

filter_prices_gt_500 = list(filter(lambda x : x > 500 , prices))
print("Prices Greater Than 500 :",filter_prices_gt_500)

filter_prices_lt_500 = list(filter(lambda x : x < 500 , prices))
print("Prices Less thatn 500 :",filter_prices_lt_500)
```
