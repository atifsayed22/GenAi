## Task 6 - Combined Utility Funtion 

```python
def process_prices(prices) : 
    discounted_prices = list(map(lambda price : price - price * 0.10 , prices ))
    filtered_prices = filter(lambda x : x > 300  , discounted_prices) 

    return discounted_prices , list(filtered_prices)  
print(process_prices([100 , 500 , 900 , 50 , 750]))
```
