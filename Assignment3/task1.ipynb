## Task 1 - Basic function  : Price After Discount 

```python
def apply_discount(price , discount = 5 ) : 
    if discount < 0 or discount > 60 : 
        print("Invalid Discount Value. Please provide a value between 0 and 60")
        return None
    discount_amount = price * (discount / 100)
    final_price = price -  discount_amount
    return final_price

print(f"Price After Discount : {apply_discount(5000 , 10)}") # Passing discount value 
print(f"Price After Discount : {apply_discount(100 )}") # Using default discount value
print(f"Price After Discount : {apply_discount(100 , 78 )}") # Passing invalid discount value
```
