## Task 7 - Mini Problem 

```python
def add_price(prices_list , price) : 
    prices_list.append(price)
    return prices_list

def get_average_price(price_list) : 
    return sum(price_list) / len(price_list)
 
def get_max_price(price_list) : 
    return max(price_list)

price_list = []
while True : 
    choice = input("Enter menu option (1. Add price, 2. Show Average price,  3. show Highest price q. Quit):")
    if choice == '1' :
        price = float(input("Enter a Price You want to add : "))
        add_price(price_list , price)
    elif choice == '2' : 
        print(get_average_price(price_list))
    elif choice == '3' : 
        print(get_max_price(price_list)) ;
    elif choice == 'q' or choice =='Q' : 
        break
    else : 
        print("Please Enter a Valid choice (1. Add price, 2. Show Average price,  3. show Highest price q. Quit) ")
        continue 

```
