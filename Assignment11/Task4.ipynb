```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 

df = pd.read_csv('sales.csv')
df['Order Date'] = pd.to_datetime(df['Order Date'])
```

```python


months = np.arange(1, 4)

def monthly_profit(year):
    return (
        df[df["Order Date"].dt.year == year]
        .groupby(df["Order Date"].dt.month)["Total Profit"]
        .sum()
        .reindex(months, fill_value=0)
    )

sales_2010 = monthly_profit(2010)
sales_2011 = monthly_profit(2011)
sales_2012 = monthly_profit(2012)

width = 0.25

plt.bar(months - width, sales_2010.values, width=width, label="2010")
plt.bar(months, sales_2011.values, width=width, label="2011")
plt.bar(months + width, sales_2012.values, width=width, label="2012")

plt.xticks(
    months,
    ["Jan", "Feb", "Mar"]
)

plt.xlabel("Month")
plt.ylabel("Total Profit")
plt.title("Profit Comparison")
plt.legend()
plt.show()


print(sales_2010)
print(sales_2011)
print(sales_2012)
```
