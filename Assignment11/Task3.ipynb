```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 

df = pd.read_csv('sales.csv')

```

```python
# 1. Create a vertical bar chart of column 

category_count = df['Item Type'].value_counts().head()

plt.xlabel("Category")
plt.ylabel("Category count")
plt.bar( category_count.index , category_count.values)
```

```python
# 2. Create a horizontal bar chart for the same data. 

category_count = df['Item Type'].value_counts().head()

plt.ylabel("Category")
plt.xlabel("Category count")

plt.barh( category_count.index , category_count.values)
```
