```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 

df = pd.read_csv('sales.csv')
```

```python
# 1. Create a  line plot showing sales trend over months 

df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Total Profit'] = df['Total Profit'].astype('int64')

sales = df[df['Order Date'].dt.year==2017]



month_sales = sales.groupby(sales['Order Date'].dt.month)['Total Revenue'].sum().head()

month_sales.index = month_sales.index.map(
    lambda x : pd.Timestamp(2017 , x , 1).strftime('%B')
)


x = month_sales.index 
y = month_sales.values


```

```python
# 2. Add : X-axis , Y-axis label and Title 

plt.title("Sales Trend over months")
plt.xlabel("Months")
plt.ylabel("Sales")

# 3. Display the plot 

plt.plot(x, y , marker='o' )
```
