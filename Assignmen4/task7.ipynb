```python
prices = {
    "Mouse" : 500 , 
    "Keyboard" : 800 , 
    "Monitor" : 7000 , 
    "Pendrive" : 400 , 
    "Camera" : 5000
}
```

```python
discount_percentage = int(input("Enter a discount percentage: "))
totalDiscount = 0 

with open("discount_report.txt" , 'w') as file : 

    for key in prices : 
        
        discounted_price = prices[key] - prices[key] * discount_percentage/100
        totalDiscount += discounted_price
        file.write(f"{key} | {prices[key]} | {discounted_price} \n")

print(f" Total Items  : {len(prices)}")
print(f" Average Discounted Price : {totalDiscount/len(prices)}")
```
