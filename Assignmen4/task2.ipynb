```python
# read the entire file using read

with open("sales_data.txt" , 'r') as file : 
    content = file.read()
  

print(content)
```

```python
#read the first line

with open("sales_data.txt" , 'r') as file :
    first_line = file.readline()

print(first_line)
```

```python
# read all the line using readlines() 

with open("sales_data.txt" , 'r') as file : 
    lines = file.readlines()


numbers = list(map(lambda x : int(x.strip()) , lines))
print(numbers)
```
