# Task 5 : Create Product Info File (user Input)

```python
product_info  = []

for i in range (1 , 4) : 
    product = input("Enter Product Name : ")
    price = int(input("Enter Product Price in Rs : "))

    product_info.append([product , price])

print(product_info)
```

```python

with open("products.txt" , 'w') as file : 
    for product in product_info : 
        file.write(product[0] + " | " + str(product[1]) + "\n")



```

```python
with open("products.txt" , 'r') as file : 
    for line in file : 
        print(line) 
```
