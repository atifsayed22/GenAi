## Task 3 Append new sales Data 


```python
data = [5000 , 2500 , 1700] 

with open("sales_data.txt" , 'a') as file : 
    for x in data : 
        file.write(str(x)+'\n')

```

```python
with open("sales_data.txt" , 'r') as file : 
    content = file.readlines()

print(content)
```
