## Task 4 Generate Summary Report from File


```python
with open("sales_data.txt" , 'r') as file : 
    content = file.readlines()

print(content)
```

```python

sales = list(map(lambda x : int(x.strip()) , content))
print(sales)
```

Total Sales 

```python
print(f"Total sales : {sum(sales)}")
```

Highest sales 

```python
print(f"Highes Sales : {max(sales)}")
```

Lowest sales 

```python
print(f"Lowest Sales : {min(sales)}")
```

Average Sales 

```python
print(f"Average Sales : {sum(sales) / len(sales)}")
```
