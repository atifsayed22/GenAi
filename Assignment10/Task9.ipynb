```python
import pandas as pd 
sales = {
    
    'Day' : ['Mon' , 'Tue' , 'Wed' , 'Thu' , 'Fri'] , 
    'Revenue' : [1200 , 1500  , 900 , 2000, 1800] 
}

df = pd.DataFrame(sales)
 
```

```python
# 1. Total Revenue 

df['Revenue'].sum()
```

```python
# 2. Average Daily Revenue 

df['Revenue'].mean()
```

```python
# 3. Day with highest Revenue 


df.loc[df['Revenue'].idxmax() , 'Day']
```

```python
# 4. Day where revenue > average 

avg = df['Revenue'].mean()

df[df['Revenue'] > avg] 
```

```python
df.plot(kind='bar', x='Day', y='Revenue')
```
