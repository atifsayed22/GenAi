```python
import pandas as pd 
students = {
    'Name' : ['amit' , 'neha' , 'Rahul' , 'Sneha' , 'Pooja'],
    'Marks' : [78 , 85 , 90 , 66 , 72 ],

    'Subject' : ['Math' , 'Math','Science','Science' , 'Math']
}

df = pd.DataFrame(students)
 
```

```python
# 1. Find average marks per subjects using groupby()

df.groupby('Subject')['Marks'].mean()
```

```python
# 2. count number of students per subjects 

df.groupby('Subject')['Marks'].count()
```

```python
# 3. Find maximum number of marks per subject 

df.groupby('Subject')['Marks'].max()
```
