```python
import pandas as pd 
students = {
    'Name' : ['amit' , 'neha' , 'Rahul' , 'Sneha' , 'Pooja'],
    'Marks' : [78 , 85 , 90 , 66 , 72 ],

    'Subject' : ['Math' , 'Math','Science','Science' , 'Math']
}

df = pd.DataFrame(students)
```

```python
# 1. Students who scored more than 75 marks 

df[df['Marks'] > 75]
```

```python
# 2. Student belonging to subject math

df[df['Subject']=='Math']
```

```python
# 3. Student who score more than average marks 

avg = df['Marks'].mean()

df[df['Marks'] > avg]
```

```python
# 4. Student who failed (marks < 70)

df[df['Marks'] < 70]
```
