```python
import pandas as pd 
marks  = pd.Series([78 , 85 , 90 , 66 , 72])
```

```python
# 1. Find Maximum , minimum , sum , mean of marks7  

print(f"Maximum marks : {max(marks)}")
print(f"Minimum marks : {min(marks)}")
print(f"sum of marks : {sum(marks)}")
print(f"Mean of marks : {sum(marks)/len(marks)}")


```

```python
# 2. Apply a lambda function to check whether each student has passed (>=70)

passed = list(filter(lambda x : x >= 70 , marks) )

print(passed)
```

```python
# 3. How many students passed 

print(f"No. of Student passed : {len(passed)}")
```
