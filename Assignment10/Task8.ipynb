```python
import pandas as pd 
students = {
    'Name' : ['amit' , 'neha' , 'Rahul' , 'Sneha' , 'Pooja'],
    'Marks' : [78 , 85 , 85 , 85 , 72 ],

    'Subject' : ['Math' , 'Math','Science','Science' , 'Math']
}

df = pd.DataFrame(students)
 
```

```python
# 1. Plot a bar grpah of student names vs marks 

df.plot(kind='bar' , x = 'Name' , y = 'Marks')
```

```python
# 2. Plot a linle Grahh of marks 

df['Marks'].plot(kind = 'line')
```

```python
# 3. Plot a histogram of marks 

df["Marks"].plot(kind = 'hist')
```
