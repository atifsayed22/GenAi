```python
def check_age(age) : 
    try : 
        if age >  120 or age < 1 : 
            raise ValueError("Age must be between 1 and 120 ")
    except ValueError as e : 
        print(e) 
    else : 
        print("Valied age")
    
age = int(input("Enter age between 1 and 120 : "))

check_age(age)
```
