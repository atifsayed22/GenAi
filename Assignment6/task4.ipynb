```python
filename = input("Enter a FileName : ")

try : 
    with open(filename , 'r') as file :
        lines = file.readlines()
except FileNotFoundError : 
    print("File Not found")

except PermissionError : 
    print("Permission denied")

else : 
    print("first 3 lines : ") 
    print("".join(lines[:3])) ; 

finally : 
    print("file operation Attemted")
```
