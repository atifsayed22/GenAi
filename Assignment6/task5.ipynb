```python
cart = []

while True : 
    price_input = input("enter a price (q to quit) : ")
    if( price_input.lower() == 'q') : 
        break 
    try : 
        price = float(price_input)

        if(price<0): 
            raise ValueError("The number is negatve")
        
        cart.append(price)
    except ValueError as e : 
        print(e)

print(f"Total Items : {len(cart)}")
print(f"Total bill : {sum(cart)}")
```
