```python
prices = [ 120 , 350 , 'abc'  , 500 , -200 , 800]
total = 0 
for price in prices : 
    try : 
        if not isinstance(price , (int , float)):
            raise TypeError("Value is not number")
        if price < 0 : 
            raise ValueError("Negative Price Not allowed")

        total += price 
    except TypeError as e : 
        print(e)
    except ValueError as e : 
        print(e)


print(f"Total : {total}" )

    

```
