```python
class Product : 

    def __init__(self , name , price , category):
        self.name = name 
        self.price = price 
        self.category = category

    def get_info (self) : 
        print("---Product Details ----")
        print(f"Product Name : {self.name} \n  Price : {self.price} \n Category : {self.category}")


product1 = Product("Laptop" , 45000 , "Electronics" ) 
product2 = Product("Mobile" , 35000 , "Electronics" ) 
product1.get_info()
product2.get_info()



```
