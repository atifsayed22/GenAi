```python
class Product : 
    def __init__(self , name , price , category):
        self.name = name 
        self.__price = price
        self.category  = category

    def get_price(self) : 
        print(self.__price) 
    def set_price(self , price) : 
        if price > 0 :
            self.__price = price 
        else : 
            print("can't set price , Price should be greater than 0 ")

product1 = Product("Laptop" , 6700 , "Electronics") 
product1.get_price()
product1.set_price(0)
product1.get_price()
```
