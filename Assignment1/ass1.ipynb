## Task 3 

```python

price_dict = {
    "Laptop": 75000,
    "Phone": 40000,
    "Keyboard": 2000,
    "Mouse": 1000,
    "Headphones": 3000,
    "Monitor": 15000
}

#add a new product and its price to the dictionary
price_dict["Tablet"] = 25000

# update the price of an existing product in the dictionary
price_dict["Laptop"] = 70000



print("Price dictionary is : " , price_dict)


# remove a product from the dictionary if exists

# product_to_remove = "Mouse"

# if product_to_remove in price_dict:
#     del price_dict[product_to_remove]
#     print(f"{product_to_remove} has been removed from the price dictionary.")
# else:
#     print(f"{product_to_remove} is not found in the price dictionary.")




total = sum(price_dict.values())
count = len(price_dict)

average = total / count

print(average)


max_price = max(price_dict.values())
min_price = min(price_dict.values())


print("Maximum price in the dictionary is : " , max_price)
print("Minimum price in the dictionary is : " , min_price)
```

## Task 4

```python

catalog = []

for product, category in zip(products, categories):
    price = price_dict[product]

    catalog.append((product, price, category))


print("Catalog of products is : " , catalog)
```
