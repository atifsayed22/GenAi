# Scatter Plot 
It is used in Bi Variate Analysis . which is used to visualize realtionship between two continuous numerical variable . it shows how on changin one numerical varibale value change the other numerical values . it's help to identitfy whether two variable are dependent or not 

eg : age vs cholestrol 

if increasing age value changes cholestrol values then there is some dependency between them 

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot  as plt 

df = pd.read_csv('depression.csv')
df.info()
```

```python
# Optimizing the memory 

df[['Age' ]] = df[['Age' ]].astype('int32')

df.info()
```

```python
age_data = df.groupby('Age')['CGPA'].mean()
top_ages = df['Age'].value_counts().head(10).sort_index()

age_data = age_data[age_data.index.isin(top_ages.index)].sort_index()

x = age_data.index 
y  = age_data.values

print(top_ages.index)
age_data.values
```

```python
plt.figure(figsize=(10,4))
plt.title("Relation between age and cgpa")
plt.xlabel("Age")
plt.grid()
plt.ylabel("Average cgpa")
plt.scatter(x , y , marker='o')
```

```python
x = np.linspace(0 , 20 , 50) 
y = np.round(np.random.random(50)*10)
y = y.astype('int32')

plt.figure(figsize=(10 , 5))
plt.scatter(x, y )
```
