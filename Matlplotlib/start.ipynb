```python
invalid = ['3.0', 'City', 'M.Com', 'M.Tech', 'ME']
df.loc[df['City'].isin(invalid), 'City'] = pd.NA

df[df['City'].isna()]
```

```python
data = df.groupby('City')['Academic Pressure'].mean().head()
x = data.index
y = data.values
plt.grid()
plt.xlabel("City")
plt.ylabel("Academic Presure")
plt.title("Academic pressure as per city")

plt.plot(x, y , marker='.' , markersize=10)
```

```python
top_cities = df['City'].value_counts().head(10).sort_values()

all_cities = df.groupby('City')['Financial Stress'].mean()

all_cities = all_cities[all_cities.index.isin(top_cities.index)].sort_index()

x = all_cities.index
y = all_cities.values

plt.figure(figsize=(10,5))
plt.plot(x, y , marker='o' , markersize= 10 , color='green'  , linestyle='-.')
plt.grid(axis='both')
plt.title("Financial Stress as per city")
plt.xlabel("City")
plt.ylabel("Financial Stress")
```

```python
sales_df = pd.read_csv('sales.csv')

sales_df['Order Date'] = pd.to_datetime(sales_df['Order Date'])
sales_df['Order Date'].dt.month_name()
```

```python
sales_df.info()
```

```python
sales_df['Total Revenue'] = sales_df['Total Revenue'].astype('int32')
sales_df['Unit Cost'] = sales_df['Unit Cost'].astype('int32')
sales_df['Total Profit'] = sales_df['Total Profit'].astype('int32')
```

```python
sales_df['Sales Channel'] = sales_df['Sales Channel'].astype('category')

sales_df.info()
```

```python
sales_df['Total Cost'] = sales_df['Total Cost'].astype('int32')
sales_df['Unit Price'] = sales_df['Unit Price'].astype('int32')
sales_df['Units Sold'] = sales_df['Units Sold'].astype('int32')
sales_df.info()
```

```python

sales_2017 = sales_df[sales_df['Order Date'].dt.year == 2017]
sales = sales_2017.groupby(sales_df['Order Date'].dt.month)['Units Sold'].sum()

sales.index  = sales.index.map(
    lambda x : pd.Timestamp(2017 , x , 1).strftime('%B')
)

```

```python
x = sales.index
y = sales.values

plt.figure(figsize=(10,5))
plt.grid()
plt.plot(x ,y , marker = 'o' , markersize = 10  ,linewidth=2.5)
plt.title("Sales as per month")
plt.xlabel("Month")
plt.ylabel("Sales")
```

```python
salesDf = sales_df[sales_df['Order Date'].dt.year==2017].set_index('Order Date').resample('ME')['Units Sold'].sum()

salesDf.index = salesDf.index.strftime('%b')

salesDf

plt.figure(figsize=(10,4))
plt.grid()
plt.title("Monthly Sales Data for year 2017 ")
plt.xlabel('Months')
plt.ylabel("Units Sold")
plt.plot(salesDf, label='Unit Sold' , marker='o' , color='green', linestyle='--')
plt.legend()
```
