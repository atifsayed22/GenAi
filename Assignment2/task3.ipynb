```python
amounts = []

while True:

    choice = input(
        "Enter menu option (1. Add amount, 2. View orders, q. Quit): "
    )

    if choice == '1':

        amount = int(input("Enter amount to add: "))
        amounts.append(amount)

    elif choice == '2':

        total_revenue = 0

        print("Amount | Discount | Final Amount")

        for order in amounts:

            if order >= 2000:
                discount_percentage = 15

            elif order >= 1500:
                discount_percentage = 10

            elif order >= 1000:
                discount_percentage = 7

            else:
                discount_percentage = 0

            discount_amount = order * discount_percentage / 100
            final_amount = order - discount_amount

            total_revenue += final_amount

            print(f"{order} | {discount_percentage}% | {final_amount}")

        print("Total Revenue:", total_revenue)

    elif choice == 'q':

        print("Exiting the program.")
        break

    else:

        print("Invalid choice. Please try again.")
```
