```python

try:
    order_amount = int(input("Enter the amount of money you want to convert: "))
    discount = 0
    if order_amount >= 2000 : 
        discount = 15
    elif order_amount >= 1500 :
         discount = 10
    elif order_amount >=1000 :
        discount = 7
    else:
        discount = 0



    discount_amount = (discount / 100) * order_amount

    final_amount = order_amount - discount_amount

    print(f"Original amount: Rs{order_amount}")
    print(f"Discount: {discount}%")
    print(f"Discount amount: Rs{discount_amount}")
    print(f"Final amount after discount: Rs{final_amount}")
except ValueError:
    print("Invalid input. Please enter a valid number.")
    exit(1)
   

```
