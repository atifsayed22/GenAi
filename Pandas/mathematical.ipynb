```python
def format_subscribers(x):
    if x >= 1_000_000_000:
        return f"{x / 1_000_000_000:.1f}B"
    elif x >= 1_000_000:
        return f"{x / 1_000_000:.1f}M"
    elif x >= 1_000:
        return f"{x / 1_000:.1f}K"
    else:
        return str(x)
```

```python
df["subscribers_readable"] = df["subscribers"].apply(format_subscribers)

df
```

```python
df[['Youtuber' ,'subscribers_readable' , 'created_year','subscribers']].head().sort_values(by='subscribers' , ascending=False)
```
