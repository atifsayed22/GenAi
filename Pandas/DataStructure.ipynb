```python
import pandas as pd 
import numpy as np
```

### Series in pandas 

series is an indivisual column in table just like  excel or sql with label 

 Each element is assigned a default integer index starting from 0

```python
data = np.array(['g' , 'e' , 'f' , 'k' , 's'])
s = pd.Series(data)

print(s)
```

```python
pd.Series([40 , 50 , 60] , index =['bread' , 'milk' ,'butter'] , name = 'price of grocery of items ' ,  dtype='int32')
```

```python
#creating series using list 

name = [['Atif' , 'Zayan'], 'Ibrahim' ,'Rayyan' , 'Umair' , 'Saud' , 'Owais']
country = ['india' , 'usa' , 'china' , 'bangladesh' , 'uae' , 'saudi arabia']
pd.Series(name )
```

```python
# creating series using numpy array 

branches = np.array([
    'It' , 'Mech' , 'Comp' , 'Extc'
])
student = ['Atif' , 'Maaz' ,'Owais' , 'Ibrahim']

pd.Series(student , branches)
```

```python
# Dicitionary to series 

marks = {
    'Maths' :[88, 98] , 
    'Chem' : [78 , 89], 
    'English' :[89 ,78] ,
}

pd.Series(marks)
```
