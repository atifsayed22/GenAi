```python
import numpy as np
marks = np.array([78 , 85 , 90 , 66 , 72 , 88 , 95 , 60])

```

1. Mean 


```python
print(np.mean(marks))
```

2. Median

```python
print(np.median(marks))
```

3. Variance 

```python
print(np.var(marks))
```

4. Standard Deviation

```python
print(np.std(marks))
```

4. Minimum & Maximum

```python
print(np.min(marks))
print(np.max(marks))
```

5. Range (max - min)

```python
print(np.max(marks) - np.min(marks))
```
