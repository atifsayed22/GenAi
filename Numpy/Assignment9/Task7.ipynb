```python
import numpy as np 
sales = np.array([1200 , 1500 , 900 , 2000 ,1800, 1700 , 1600])

print(np.size(sales))
```

```python
# 1. Total Weekly sales 

total_sales = np.sum(sales)
print(total_sales)
```

```python
# 2. Average Daily Sales

avg_daily_sales = np.mean(sales)
print(f"{avg_daily_sales:.2f}")
```

```python
# 3. Highest and lowest sales day 

highest_sales = np.max(sales) 
lowest_sales = np.min(sales) 


print(f"Highest Sales Day : {np.argmax(sales)}")
print(f"Lowest Sales Day : {np.argmin(sales)}")
```

```python
# 4. Standar Deviation of Sales

print(f"Standar Deviation of sales: {np.std(sales):.2f}")
```

```python
# 5. Identify Days where sales were abouve average 

avg = np.mean(sales)
print(avg) 

print(f"Days where sales are  above average :")


for i , sale  in enumerate(sales) : 
    if(sale>avg) : 
        print(f"Day {i+1} ")
```
