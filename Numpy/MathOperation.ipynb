# Important Function in Numpy for data science and machine learning 


```python
# Max() calculate the max value in the given matrix or array 

print(np.max(arr_1))
```

axis ---> it provide us the max or min value arround a specific row or column 

axis = 0 ----> column wise 
axis = 1 ----> row wise

```python
print(arr_1)
print(np.max(arr_1, axis=0)) # give the max value in the cloumn 
print(np.max(arr_1, axis=1)) # give the max value in the row 
```

sum() ---> calculate the sum of all the value in the arrya or matrix 

```python
np.sum(arr_1 , axis=1) # sum of value row wise 
```

prod() ---> calculate the product of the all value in tht array or matrix 

```python
print(np.prod(arr_1 ))
print(np.prod(arr_1 , axis=0)) # column
print(np.prod(arr_1  , axis = 1) ) # row 
```

# Stastical function in NumPY 

mean() ---> mean of all values in the array 


```python
r = np.mean(arr_1 , axis=1) # mean value for every row 
c = np.mean(arr_1 , axis = 0 ) # mean value for every column 
a = np.mean(arr_1) # mean value of entire matrix or array
print(arr_1)
print(f"mean value of every row :{r}")
print(f"mean value of every column :{c}")
print(f"mean value of entire matrix:{a}")
```

```python
sd_array = np.array([[70 , 70 , 70 , 70 , 70]  , [30, 50, 90, 90, 110] , [0 , 0 , 0  , 0 , 0 ]])

print(sd_array)

m = np.mean(sd_array) 
print(f"Mean : {m}")
sd = np.std(sd_array , )
print(f"std : {sd}")

md = np.median(sd_array , axis=1).astype('bool')
print(md)
print(md.dtype)
```

standard deviation std() . calculate the sd for every values in the element . sd tell how spread a value is from the mean . 


variance() ---- > variance calculate the how spread out the data points from the mean . by taking average of squared diffrence between each data and mean 


var = (mean - x)  ^ 2 / n 

```python
v = np.var(sd_array , axis = 1 ).astype("float")
s = np.std(sd_array , axis = 1 ).astype("float")
print(v )
print(s)
```

# Dot Products

```python
matrix_1 = np.round(np.random.random((3 , 4)) * 50 ).astype("int32")
matrix_2 = np.round(np.random.random((4 , 4)) * 50 ).astype("int32")
print(matrix_1)
print(matrix_2)

dp = np.dot(matrix_1 , matrix_2)
print(dp)
```

log values -- -> log() calculate the log values of each and every values 

```python
lg = np.log(matrix_1)
print(matrix_1)
print(lg )

np.log([10])
```

round , ciel , floor 

round --- > goes to nearest integer value 
ciel --- > goest towared smallest integer which is greater than or equal  value eg : 1.3 --- > 2 
floor --- . goes downwards ege 1.8 --- > 1 return the largest integer which is less than or equal to the given numner 

```python
r  = np.round([10.8 , 9.5 , 12.3])
c  = np.ceil([10.8 , 9.5 , 12.3])
f  = np.floor([10.8 , 9.5 , 12.3])
print(f'round : ;{r}')
print(f'ceil : ;{c}')
print(f'floor : ;{f}')
```

transpose: np.transpose(arr)   arr[i][j] ---> arr[j][i]



```python
mat = np.round(np.random.random((3 , 4)) * 20 ).astype('int32')
print(mat)

print(np.transpose(mat) )
```
