# What is NumPy?

NumPy (Numerical Python) is a Python library used for fast and efficient
numerical computing.

It provides a multidimensional array called `ndarray` and many functions
for performing mathematical and scientific operations on numerical data.

## Why use NumPy instead of Python's built-in data structures?

Python lists are general-purpose data structures. They are flexible and can
store different types of objects, but they are not specifically optimized
for large-scale numerical computation.

NumPy arrays are designed specifically for numerical data. They generally
use memory more efficiently and allow mathematical operations to be
performed much faster through vectorized operations.

## Why is NumPy efficient?

Much of NumPy's core functionality is implemented in compiled low-level
code such as C. NumPy provides a Python interface on top of this
implementation, allowing us to perform efficient numerical operations
using simple Python syntax.
